# 실습1 (Iris)

## 1. Decision Tree



```python
import pandas as pd

from sklearn.model_selection import train_test_split

from sklearn.tree import DecisionTreeClassifier

from sklearn.metrics import classification_report
```

```python
df = pd.read_csv("/content/drive/MyDrive/Colab Notebooks/iris.csv")
```

```python
X = df.drop("species", axis=1)
y = df["species"]
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)
```

```python
model = DecisionTreeClassifier()
model.fit(X_train, y_train)
```

```python
pred_dt = model.predict(X_test)
print(classification_report(y_test, pred_dt))
```

## 2. Random Forest

```python
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report


df = pd.read_csv("/content/drive/MyDrive/Colab Notebooks/iris.csv")

X = df.drop("species", axis=1)
y = df["species"]

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)

model = RandomForestClassifier()

model.fit(X_train, y_train)

pred_rf = model.predict(X_test)

print(classification_report(y_test, pred_rf))
```

## 3. Logistic Regression

```python
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report


df = pd.read_csv("/content/drive/MyDrive/Colab Notebooks/iris.csv")

X = df.drop("species", axis=1)
y = df["species"]

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)

model = LogisticRegression(max_iter=200)

model.fit(X_train, y_train)

pred_lr = model.predict(X_test)

print(classification_report(y_test, pred_lr))
```

# 실습3 (Diabetes)

## 1.Decision Tree

```python
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeRegressor
from sklearn.metrics import mean_squared_error


df = pd.read_csv("/content/drive/MyDrive/Colab Notebooks/diabetes.csv")

# Outcome 제거
df = df.drop("Outcome", axis=1)

# BMI를 정답(y)으로 분리
X = df.drop("BMI", axis=1)
y = df["BMI"]

# 학습용 / 테스트용 데이터 분리
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2
)

# Decision Tree
model = DecisionTreeRegressor()

model.fit(X_train, y_train)

# 예측
pred_dt = model.predict(X_test)

# 결과
print("평균제곱오차:", mean_squared_error(y_test, pred_dt))
```

## 숙제2. 회귀(Wine)

## 2-1. Decision Tree

```python
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeRegressor
from sklearn.metrics import mean_squared_error


df = pd.read_csv("/content/drive/MyDrive/Colab Notebooks/wine.csv")

# Wine 제거
df = df.drop("Wine", axis=1)

# Alcohol을 정답(y)으로 분리
X = df.drop("Alcohol", axis=1)
y = df["Alcohol"]

# 학습용 / 테스트용 데이터 분리
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2
)

# Decision Tree Regression
model = DecisionTreeRegressor()

model.fit(X_train, y_train)

# 예측
pred_dt = model.predict(X_test)

# 평균제곱오차
print("평균제곱오차:", mean_squared_error(y_test, pred_dt))
```
