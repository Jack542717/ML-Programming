## 실습 2. Auto MPG

### Decision Tree

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeRegressor
from sklearn.metrics import mean_squared_error


# 컬럼 이름 설정
columns = [
    "mpg",
    "cylinders",
    "displacement",
    "horsepower",
    "weight",
    "acceleration",
    "model_year",
    "origin",
    "car_name"
]


# 데이터 불러오기
df = pd.read_csv(
    "/content/drive/MyDrive/Colab Notebooks/auto-mpg.data",
    sep=r"\s+",
    names=columns
)


# horsepower의 '?' 값을 숫자로 변환
# 숫자로 변환할 수 없는 '?'는 NaN으로 처리
df["horsepower"] = pd.to_numeric(
    df["horsepower"],
    errors="coerce"
)


# 결측값 제거
df = df.dropna()


# X, y 분리
# mpg를 예측하고, 자동차 이름은 입력값에서 제외
X = df.drop(["mpg", "car_name"], axis=1)
y = df["mpg"]


# 학습 데이터 / 테스트 데이터 분리
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.2
)


# Decision Tree 모델
model = DecisionTreeRegressor()

model.fit(X_train, y_train)


# 예측
pred_dt = model.predict(X_test)


# 결과 출력
print(mean_squared_error(y_test, pred_dt))
```

### Random Forest

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_squared_error


# 컬럼 이름 설정
columns = [
    "mpg",
    "cylinders",
    "displacement",
    "horsepower",
    "weight",
    "acceleration",
    "model_year",
    "origin",
    "car_name"
]


# 데이터 불러오기
df = pd.read_csv(
    "/content/drive/MyDrive/Colab Notebooks/auto-mpg.data",
    sep=r"\s+",
    names=columns
)


# horsepower의 '?' 값을 결측값으로 처리
df["horsepower"] = pd.to_numeric(df["horsepower"], errors="coerce")


# 결측값 제거
df = df.dropna()


# X, y 분리
X = df.drop(["mpg", "car_name"], axis=1)
y = df["mpg"]


# 학습 데이터 / 테스트 데이터 분리
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.2
)


# Random Forest 모델
model = RandomForestRegressor()

model.fit(X_train, y_train)


# 예측
pred_rf = model.predict(X_test)


# 결과 출력
print(mean_squared_error(y_test, pred_rf))
```

### KNN

```python
# 5. KNN Regression

import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsRegressor
from sklearn.metrics import mean_squared_error


# 컬럼 이름 설정
columns = [
    "mpg",
    "cylinders",
    "displacement",
    "horsepower",
    "weight",
    "acceleration",
    "model_year",
    "origin",
    "car_name"
]


# 데이터 불러오기
df = pd.read_csv(
    "/content/drive/MyDrive/Colab Notebooks/auto-mpg.data",
    sep=r"\s+",
    names=columns
)


# horsepower의 '?' 값을 결측값으로 처리
df["horsepower"] = pd.to_numeric(
    df["horsepower"],
    errors="coerce"
)


# 결측값 제거
df = df.dropna()


# X, y 분리
X = df.drop(["mpg", "car_name"], axis=1)
y = df["mpg"]


# 학습 데이터 / 테스트 데이터 분리
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.2
)


# KNN 모델
model = KNeighborsRegressor()

model.fit(X_train, y_train)


# 예측
pred_knn = model.predict(X_test)


# 결과 출력
print(mean_squared_error(y_test, pred_knn))
```
